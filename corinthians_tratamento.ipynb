```python
partidas["data"] = pd.to_datetime(
    partidas["data"],
    format="%d/%m/%Y",
    errors="coerce"
)
```

```python
partidas[['data']].head()
```

```python
part_cor = partidas[(partidas['mandante'] =='Corinthians')|(partidas['visitante'] == 'Corinthians')].copy()
part_cor
```

```python
print('Quantidade de Partidas  - Corinthians ( entre 2003 - 2025): ',len(part_cor))

display(part_cor[['ID','data','mandante','visitante','mandante_Placar', 'visitante_Placar']].head())
```

```python
ids_cor = part_cor['ID']
```

```python
estatic_cor = estatic[(estatic['partida_id'].isin(ids_cor)) & (estatic['clube']=='Corinthians')].copy()

gols_cor = gols[(gols['partida_id'].isin(ids_cor)) & (gols['clube']=='Corinthians')].copy()

cartao_cor = cartao[(cartao['partida_id'].isin(ids_cor)) & (cartao['clube']=='Corinthians')].copy()
```

```python
print("Partidas:", part_cor.shape)
print("Estatísticas:", estatic_cor.shape)
print("Gols:", gols_cor.shape)
print("Cartões:", cartao_cor.shape)
```

```python
print(
    "Partidas com estatísticas:",
    estatic_cor["partida_id"].nunique()
)

print(
    "Partidas com autores de gols:",
    gols_cor["partida_id"].nunique()
)

print(
    "Partidas com cartões registrados:",
    cartao_cor["partida_id"].nunique()
)
```

```python
part_cor['temporada'] = part_cor['data'].dt.year

jogos_finais_2020 = ((part_cor['data'].dt.year ==2021) & (part_cor['data'].dt.month <= 10))

part_cor.loc[jogos_finais_2020, 'temporada'] = 2020
part_cor[ 'temporada'] = part_cor['temporada'].astype(int)
```

```python
part_cor.groupby('temporada').size()
```

```python
part_cor['local'] = np.where(part_cor['mandante'] == 'Corinthians', 'Mandante', 'Visitante')

part_cor['adversario'] = np.where(
    part_cor['mandante'] == 'Corinthians',
    part_cor['mandante_Placar'],
    part_cor['visitante_Placar']
)

part_cor['placar'] = np.where(
    part_cor['mandante'] == 'Corinthians',
    part_cor['mandante_Placar'],
    part_cor['visitante_Placar']
)

part_cor["gols_marcados"] = np.where(
    part_cor["mandante"] == "Corinthians",
    part_cor["mandante_Placar"],
    part_cor["visitante_Placar"]
)

part_cor['gols_sofridos'] = np.where(
    part_cor['mandante'] == 'Corinthians',
    part_cor['visitante_Placar'],
    part_cor['mandante_Placar']
)


part_cor['saldo_gols'] = (
    part_cor['gols_marcados'] -
    part_cor['gols_sofridos']
)

condicoes = [
    part_cor['gols_marcados'] >
    part_cor['gols_sofridos'],

    part_cor['gols_marcados'] <
    part_cor['gols_sofridos']
  ]
resultados = ['Vitoria', 'Derrota']

part_cor['resultado'] = np.select(
    condicoes,
    resultados,
    default='Empate'
)

```

```python
part_cor['pontos'] = part_cor['resultado'].map({
    'Vitoria': 3,
    'Derrota': 0,
    'Empate': 1
})
```

```python
part_cor['tecnico'] = np.where(
    part_cor['mandante'] == 'Corinthians',
    part_cor['tecnico_mandante'],
    part_cor['tecnico_visitante']
)

part_cor['formacao'] = np.where(
    part_cor['mandante'] == 'Corinthians',
    part_cor['formacao_mandante'],
    part_cor['formacao_visitante']
)
```

```python
colunas_part = [
    'ID',
    'temporada',
    'rodata',
    'data',
    'hora',
    'adversario',
    'local',
    'arena',
    'gols_marcados',
    'gols_sofridos',
    'saldo_gols',
    'resultado',
    'pontos',
    'tecnico',
    'formacao',
    'arrecadacao'

]

base_part = part_cor[colunas_part].copy()
```

```python
display(base_part.head())
```

```python
resumo_geral = {
    "Jogos": len(base_part),
    "Vitórias": (base_part["resultado"] == "Vitoria").sum(),
    "Empates": (base_part["resultado"] == "Empate").sum(),
    "Derrotas": (base_part["resultado"] == "Derrota").sum(),
    "Gols marcados": base_part["gols_marcados"].sum(),
    "Gols sofridos": base_part["gols_sofridos"].sum(),
    "Pontos": base_part["pontos"].sum()
}

pd.Series(resumo_geral)
```

```python
pontos_possiveis = len(base_part) * 3

aproveitamento = (base_part['pontos'].sum()/pontos_possiveis) *100

print(f'Aproveitamento geral: {aproveitamento:.2f}')
```

```python
display(
    base_part[
        [
            "temporada",
            "data",
            "adversario",
            "local",
            "gols_marcados",
            "gols_sofridos",
            "resultado",
            "pontos"
        ]
    ].head(10)
)
```

```python
coluna_estat = [

    'partida_id',
    'chutes',
    'chutes_no_alvo',
    'posse_de_bola',
    'precisao_passes',
    'passes',
    'faltas',
    'cartao_amarelo',
    'cartao_vermelho',
    'impedimentos',
    'escanteios'
]

estatic_cor_limpa = estatic_cor[coluna_estat].copy()
```

```python
colunas_percentuais = ['posse_de_bola','precisao_passes']

for coluna in colunas_percentuais:
    estatic_cor_limpa[coluna] = (estatic_cor_limpa[coluna].astype(str).str.replace('%','', regex=False)
    .str.replace(',', '.', regex=False))
    estatic_cor_limpa[coluna] = pd.to_numeric(estatic_cor_limpa[coluna], errors='coerce')


estatic_cor_limpa.dtypes
```

```python
estatic_cor_limpa["partida_id"].duplicated().sum()
```

```python
base_part = base_part.merge(estatic_cor_limpa, how='left',left_on='ID', right_on='partida_id',validate='one_to_one')

base_part.drop(columns='partida_id',inplace=True)
```

```python
print("Linhas:", base_part.shape)
print("Colunas:", base_part.columns.tolist())
```

```python
display(
    base_part[
        [
            "temporada",
            "data",
            "adversario",
            "resultado",
            "gols_marcados",
            "chutes",
            "chutes_no_alvo",
            "posse_de_bola",
            "passes",
            "precisao_passes"
        ]
    ].tail(10)
)
```

```python
base_part['estatisticas_disponiveis'] = (base_part['posse_de_bola'].notna()| (base_part['passes'] > 0) | (base_part['chutes']>0))

disponibilidade = (base_part.groupby('temporada').agg(partidas=('ID','count'), partida_com_estatistica=('estatisticas_disponiveis','sum')).reset_index( ))

display(disponibilidade)
```

```python
#aproveitamento por partida
base_part['aproveitamneto_partida'] = (base_part['pontos']/3)*100

```

```python
base_part["precisao_finalizacao"] = np.where(
    base_part["chutes"] > 0,
    (
        base_part["chutes_no_alvo"] /
        base_part["chutes"]
    ) * 100,
    np.nan
)
```

```python
base_part['conversao_chutes'] = np.where(
    base_part['chutes'] > 0,
    (
        base_part['gols_marcados'] /
        base_part['chutes']
    )*100,
    np.nan
)
```

```python
base_part["conversao_no_alvo"] = np.where(
    base_part["chutes_no_alvo"] > 0,
    (
        base_part["gols_marcados"] /
        base_part["chutes_no_alvo"]
    ) * 100,
    np.nan
)
```

```python
resumo_temporadas = (
    base_part
    .groupby("temporada")
    .agg(
        jogos=("ID", "count"),
        pontos=("pontos", "sum"),
        gols_marcados=("gols_marcados", "sum"),
        gols_sofridos=("gols_sofridos", "sum"),
        media_chutes=("chutes", "mean"),
        media_posse=("posse_de_bola", "mean"),
        media_passes=("passes", "mean")
    )
    .reset_index()
)

resumo_temporadas["aproveitamento"] = (
    resumo_temporadas["pontos"] /
    (resumo_temporadas["jogos"] * 3)
) * 100

display(resumo_temporadas.tail(10).round(2))
```

```python
base_gols = gols[
    (gols['partida_id'].isin(base_part['ID'])) &
    (gols['clube'] == 'Corinthians')
].copy()
```

```python
print(base_gols.shape)
display(base_gols.head())
```

```python
info_partidas = base_part[
    [
        "ID",
        "temporada",
        "data",
        "adversario",
        "local",
        "resultado"
    ]
].copy()
```

```python
base_gols = base_gols.merge(
    info_partidas,
    how="left",
    left_on="partida_id",
    right_on="ID",
    validate="many_to_one"
)
```

```python
base_gols[ 'minuto_original'] = base_gols['minuto'].astype(str)

def converter_minuto(minuto):
  if pd.isna(minuto):
    return np.nan

  minuto = str(minuto).strip()

  if minuto.lower() == 'nan':
    return np.nan

  if '+' in minuto:
    partes = minuto.split('+')

    try:
      return int (partes[0]) + int(partes[1])
    except:
      return np.nan
  try:
    return int(float(minuto))

  except:
    return np.nan

```

```python
    base_gols['minuto_numerico'] = (
        base_gols['minuto']
        .apply(converter_minuto)
    )
```

```python
faixas = [ 0, 15 , 30, 45, 60,75 ,90, float('inf')]

nomes_faixas = [
    '0-15',
    '16-30',
    '31-45',
    '46-60',
    '61-75',
    '76-90',
    '90+'
]

base_gols["intervalo_gol"] = pd.cut(

    base_gols["minuto_numerico"],

    bins=faixas,

    labels=nomes_faixas,

    include_lowest=True

)

base_gols['intervalo_gol'].value_counts().sort_index()
```

```python
base_gols['tipo_de_gol'] = (
    base_gols['tipo_de_gol']
    .fillna('Gol normal')
    .replace({
        'Penalty': 'Pênalti',
        'Own-goal': 'Gol contra'
    })
    )
```

```python
base_cartoes = cartao[
    (cartao['partida_id'].isin(base_part["ID"])) &
    (cartao['clube'] == 'Corinthians')
].copy()

print(base_cartoes.shape)
display(base_cartoes.head())
```

```python
base_cartoes['cartao'].value_counts(dropna=False)
```

```python
# Recria a base de cartões a partir da tabela original
base_cartoes = cartao[
    (cartao["partida_id"].isin(base_part["ID"])) &
    (cartao["clube"] == "Corinthians")
].copy()

# Informações das partidas
info_partidas = base_part[
    [
        "ID",
        "temporada",
        "data",
        "adversario",
        "local",
        "resultado"
    ]
].copy()

# Adiciona os dados da partida em cada cartão
base_cartoes = base_cartoes.merge(
    info_partidas,
    how="left",
    left_on="partida_id",
    right_on="ID",
    validate="many_to_one"
)

# Remove o ID duplicado
base_cartoes.drop(columns="ID", inplace=True)
```

```python
def classificar_cartao(cartao):
  cartao = str(cartao).lower()

  if 'vermelho' in cartao:
    return 'Vermelho'

  if 'amarelo' in cartao:
    return 'Amarelo'

  return 'Outro'

base_cartoes["tipo_cartao"] = (

    base_cartoes["cartao"]

    .apply(classificar_cartao)

)

base_cartoes["tipo_cartao"].value_counts()
```
